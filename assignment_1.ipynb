[5] Find the mean of values of dictionary.

````python
my_dict = {
    'marks1': 23,
    'marks2': 125,
    'marks3': 43,
    'marks4': 13,
    'marks5': 39
}

average = 0

for key in my_dict:
    average += my_dict[key]

average /= len(my_dict)

print(average)
````

[6] `my_dict = {'name': ['Yash', 'Neel', 'Dev'], 'rollno': [11, 12, 13], 'marks': [78, 56, 98]}`<br/>
Perform the following operations on above dictionary:<br/>
a) Display name as Dev<br/>
b) Display 12 roll no<br/>
c) Display greatest marks

````python
my_dict = {'name': ['Yash', 'Neel', 'Dev'], 'rollno': [11, 12, 13], 'marks': [78, 56, 98]}

# A: Display name as Dev
print(my_dict['name'][2])

# B: Display 12 roll no
print(my_dict['rollno'][1])

# C: Display greatest marks
greatest_marks = my_dict['marks'][0]

for i in range(1, len(my_dict['marks'])):
    if my_dict['marks'][i] > greatest_marks:
        greatest_marks = my_dict['marks'][i]

print(greatest_marks)
````

[7] Calculate frequency of letters occuring in a given string.

````python
string = "adnan"
frequencies = {}

for c in string:
    if c in frequencies:
        frequencies[c] += 1
    else:
        frequencies[c] = 1

print(frequencies)
````

[9] Suppose a date is represented as a tupe (d, m, y). Write a program to create two date tuples with values taken from the user and find the number of days between the two dates as an output.

````python
days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

d1 = int(input("Enter day of the first date: "))
m1 = int(input("Enter month of the first date: "))
y1 = int(input("Enter year of the first date: "))

d2 = int(input("\nEnter day of the second date: "))
m2 = int(input("Enter month of the second date: "))
y2 = int(input("Enter year of the second date: "))

if (y1, m1, d1) > (y2, m2, d2):
    from_date = (d2, m2, y2)
    to_date = (d1, m1, y1)
else:
    from_date = (d1, m1, y1)
    to_date = (d2, m2, y2)

d1, m1, y1 = from_date
d2, m2, y2 = to_date

days = 0

for y in range(y1, y2):
    if (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0):
        days += 366
    else:
        days += 365

for m in range(1, m1):
    days -= days_in_month[m - 1]
    if m == 2 and ((y1 % 4 == 0 and y1 % 100 != 0) or (y1 % 400 == 0)):
        days -= 1

days -= d1

for m in range(1, m2):
    days += days_in_month[m - 1]
    if m == 2 and ((y2 % 4 == 0 and y2 % 100 != 0) or (y2 % 400 == 0)):
        days += 1

days += d2

print("\nNumber of days between the dates:", days)
````
