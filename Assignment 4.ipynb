```python
matrix1=[
    [1,2,3,4],
    [5,6,7,8]
]
matrix2=[
    [2,4,6,8],
    [3,6,9,12]
]
result=[
    [0,0,0,0],
    [0,0,0,0]
]
for i in range(len(matrix1)):
    for j in range(len(matrix1[0])):
        result[i][j]=matrix1[i][j]+matrix2[i][j]
print("sum of ",matrix1,"+",matrix2,"=",result)
```
