```python
import re

#Check if the strig starts with "The" and ends with "Spain":
txt="The rain in Spain"
x=re.search("^The.*Spain$",txt)

if x:
    print("YES!, We have a match")
else:
    print("No match")
```

```python
import re

def Check_string(s):
    pattern="^[A-Za-z0-9]+$" #^=start, $=end,+=one or more
    return bool(re.match(pattern, s))

#example:
s1="hello123"
s2="hello_123" #contains special chara
print(Check_string(s1))
print(Check_string(s2))
```

```python
def check_string(s):
    return s.isalnum()
#eg:
s1="Hello123"
s2="Hello 123!" #contains space
print(Check_string(s1))
print(Check_string(s2))
```
